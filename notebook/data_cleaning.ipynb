# Biased Data Handling
- Threat categor data is very very less, so for the better model performance need more data.
- For the abusive category data, is not sufficient, so need more data for better handling.

```python
biased_df = pd.read_csv("../data/raw/jigsaw-unintended-bias-train-processed-seqlen128.csv")
```

```python
biased_df = biased_df[["comment_text","toxic", "threat", "toxicity_annotator_count"]].copy()
```

```python
print(biased_df.columns.tolist())
print(biased_df.shape)
```

```python
biased_df.head(4)
```

```python
biased_df.head(4)
```

```python
# 1. Extra threats from the big file
ub_threats = biased_df[biased_df["threat"] >= 0.5][["comment_text"]].copy()
ub_threats["category"] = "threat"
```

```python
# 2. Leakage fix: normal and abusive from the same file
ub_normal = biased_df[(biased_df["toxic"] < 0.1) & (biased_df["threat"] < 0.1)] \
    .sample(n=30000, random_state=42)[["comment_text"]]
ub_normal["category"] = "normal"
ub_abusive = biased_df[(biased_df["toxic"] >= 0.5) & (biased_df["threat"] < 0.5)] \
    .sample(n=10000, random_state=42)[["comment_text"]]
ub_abusive["category"] = "abusive"
```

```python
# 3. Combine with the main file and remove duplicates
combined = pd.concat(
    [df[["comment_text", "category"]], ub_threats, ub_normal, ub_abusive],
    ignore_index=True
)
combined = combined.drop_duplicates(subset="comment_text")

print(len(ub_threats), "extra threats")
print(combined["category"].value_counts())
```

```python
combined.head(4)
```

```python
import re
from sklearn.model_selection import train_test_split
```

```python
# 1. Light cleaning
def clean(t):
    t = re.sub(r"http\S+|www\.\S+", " ", t)          # URLs
    t = re.sub(r"\b\d{1,3}(\.\d{1,3}){3}\b", " ", t)  # IP addresses
    t = re.sub(r"\s+", " ", t)                        # newlines, extra spaces
    return t.strip()
```

```python
combined["text"] = combined["comment_text"].astype(str).map(clean)
combined = combined[combined["text"].str.len() > 0]
```

```python
combined.head(5)
```

```python
# 2. Stratified split: 80% train, 10% val, 10% test
train, temp = train_test_split(combined, test_size=0.2,
                               stratify=combined["category"], random_state=42)
val, test = train_test_split(temp, test_size=0.5,
                             stratify=temp["category"], random_state=42)
```

```python
# 3. Downsample normal in training only
normal = train[train["category"] == "normal"].sample(n=60000, random_state=42)
train_bal = pd.concat([normal, train[train["category"] != "normal"]])
train_bal = train_bal.sample(frac=1, random_state=42)
```

```python
# 4. Save
for name, d in [("train", train_bal), ("val", val), ("test", test)]:
    d[["text", "category"]].to_csv(f"{name}.csv", index=False)
```

```python
print("TRAIN:\n", train_bal["category"].value_counts())
print("VAL:\n", val["category"].value_counts())
print("TEST:\n", test["category"].value_counts())
```

```python
from scipy.sparse import hstack
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
```

```python
train = pd.read_csv("train.csv")
val = pd.read_csv("val.csv")
```

```python
# 1. Features
word_vec = TfidfVectorizer(ngram_range=(1, 2), max_features=100000,
                           sublinear_tf=True, lowercase=True)
char_vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5),
                           max_features=100000, sublinear_tf=True)

X_train = hstack([word_vec.fit_transform(train["text"]),
                  char_vec.fit_transform(train["text"])]).tocsr()
X_val = hstack([word_vec.transform(val["text"]),
                char_vec.transform(val["text"])]).tocsr()
```

```python
# 2. Model with class weights
model = LogisticRegression(max_iter=1000, class_weight="balanced", C=4)
model.fit(X_train, train["category"])
```

```python
# 3. Evaluate on validation
preds = model.predict(X_val)
print(classification_report(val["category"], preds, digits=3))
print(confusion_matrix(val["category"], preds,
                       labels=["normal", "abusive", "threat"]))
```

```python
def predict(texts):
    X = hstack([word_vec.transform(texts), char_vec.transform(texts)])
    for t, p in zip(texts, model.predict(X)):
        print(f"{p:8} | {t}")

predict([
    "Please find the report attached, deadline is Friday.",
    "Sorry to hear your father passed away, my condolences.",
    "You are a useless idiot.",
    "If you send that again I will kill you.",
    "We need to kill this project before the budget review.",
])
```

# Fine Tuning

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score
```

### Threshold tuning

```python
proba = model.predict_proba(X_val)
classes = list(model.classes_)
t_idx = classes.index("threat")
y_true = (val["category"] == "threat").values

for t in [0.5, 0.4, 0.3, 0.2, 0.15, 0.1]:
    y_pred = proba[:, t_idx] >= t
    print(f"threshold {t:.2f} | recall {recall_score(y_true, y_pred):.3f} "
          f"| precision {precision_score(y_true, y_pred):.3f}")
```

```python
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import ComplementNB
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import classification_report, f1_score
```

```python


models = {
    "LogReg":       LogisticRegression(max_iter=1000, class_weight="balanced", C=4),
    "LinearSVC":    LinearSVC(class_weight="balanced", C=0.5),
    "ComplementNB": ComplementNB(alpha=0.3),
    "SGD":          SGDClassifier(loss="modified_huber", class_weight="balanced",
                                  max_iter=50, random_state=42),
}

for name, m in models.items():
    m.fit(X_train, train["category"])
    p = m.predict(X_val)
    rep = classification_report(val["category"], p, output_dict=True)
    print(f"{name:13} | threat recall {rep['threat']['recall']:.3f} "
          f"| threat precision {rep['threat']['precision']:.3f} "
          f"| macro F1 {f1_score(val['category'], p, average='macro'):.3f}")
```
